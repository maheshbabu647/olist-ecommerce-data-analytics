```python
from pathlib import Path

project_root = Path.cwd().parents[1]
raw_folder = project_root / "data" / "raw"

print("Project root:", project_root)
print("Raw folder exists:", raw_folder.exists())

files = list(raw_folder.glob("*.csv"))

print("CSV files found:", len(files))

print([f.name for f in files])
```

```python
import pandas as pd

orders_path = raw_folder / "olist_orders_dataset.csv"
orders = pd.read_csv(orders_path)

print("shape : ", orders.shape)
print('\nData types : ')
print(orders.dtypes)
print("\nSample records: ")
print(orders.head())
```

```python
orders.isnull().sum()
```

```python
(orders.isnull().sum() / len(orders)) * 100
```

```python
(orders.isnull().sum() / len(orders)) * 100
```

```python
orders.groupby('order_status')[
   ["order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date"
   ]
].apply(lambda group: group.isnull().sum())
```

```python
orders.groupby("order_status")[
    "order_delivered_customer_date"
].apply(lambda x: x.isnull().mean() * 100)
```

```python
orders['order_status'].value_counts()
```

```python
total_bytes = sum(
    f.stat().st_size
    for f in raw_folder.rglob("*")
    if f.is_file()
)

print(f"Total size: {total_bytes / (1024**2):.2f} MB")
print(f"Total size: {total_bytes / (1024**3):.2f} GB")
```
